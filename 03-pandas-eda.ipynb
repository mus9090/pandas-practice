How to read a boxplot 

Line inside the box: median price.

Box: middle 50% of prices.

Whiskers and points: spread and potential outliers.

```python
bnb.groupby("neighbourhood_group")["price"].mean().sort_values().plot(kind="bar", color="blue", edgecolor="black")
plt.xlabel("Neighbourhood Group")
plt.ylabel("Average Price")
```

```python
bnb1=bnb.copy()
bnb1["rating"]=bnb1["rating"].sort_values(ascending=False)
bnb1["rating"].head()
```

```python
bnb["rating"]=pd.to_numeric(bnb["rating"], errors="coerce")
```

```python
bnb.nlargest(20, 'rating')
```

```python
bnb.groupby("neighbourhood_group")["rating"].mean().sort_values().plot(kind="barh", color="blue", edgecolor="black")
plt.ylabel("Neighbourhood Group")
plt.xlabel("Average Rating")  
plt.show()
```

```python
bnb["rating"].plot(kind="hist", bins=50, color="blue", edgecolor="black")
plt.xlabel("Rating")
plt.ylabel("Frequency")
```

```python
sns.boxplot(x="neighbourhood_group", y="rating", data=bnb)
plt.xlabel("Neighbourhood Group")
plt.ylabel("Rating")
plt.show()
```

```python
sns.scatterplot(x="price", y="rating", data=bnb[bnb["price"]<=bnb["price"].quantile(0.99)])
plt.xlabel("Price")
plt.ylabel("Rating")    
plt.show()
```
