How to read a boxplot 

Line inside the box: median price.

Box: middle 50% of prices.

Whiskers and points: spread and potential outliers.

```python
bnb.groupby("neighbourhood_group")["price"].mean().sort_values().plot(kind="bar", color="blue", edgecolor="black")
plt.xlabel("Neighbourhood Group")
plt.ylabel("Average Price")
```
