```python
import pandas as pd
customer=pd.read_excel('./data/Customer Call List.xlsx')
customer.head()
```

```python
customer[["Paying Customer","Do_Not_Contact"]]=customer[["Paying Customer","Do_Not_Contact"]].replace({"Y":"Yes","N":"No"})
customer
```

```python
customer["Last_Name"]=customer["Last_Name"].str.strip("._/")
customer
```

```python
customer["Phone_Number"]=customer["Phone_Number"].fillna("")
customer
```

```python
customer["Phone_Number"]=customer["Phone_Number"].str.replace("[^0-9a-zA-Z`]", "", regex=True).fillna("")
customer
```

```python
customer=customer.replace({"Na":"", "N/a":""})
customer=customer.fillna("")
customer
```

```python
customer=customer[customer["Do_Not_Contact"]!="Yes"]
customer=customer[customer["Phone_Number"]!=""]
customer
```

```python
customer.drop(columns=["Not_Useful_Column"], inplace=True)
customer
```

```python
customer=customer.reset_index(drop=True)
customer
```

```python
customer["Phone_Number"]=customer["Phone_Number"].apply(lambda x: x[0:3]+"-"+ x[3:6]+"-"+ x[6:10])
customer
```
