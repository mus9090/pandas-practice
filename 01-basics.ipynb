```python
bios["height_category"]=bios["height_cm"].apply(lambda x: "Short" if x<165 else ("Avg" if x<175 else "Tall"))
bios.head(20)
```

```python
def categorize_atheletes(x):
    if x["height_cm"]<175 and x["weight_kg"]<70:
        return "Light weight"
    elif x["height_cm"]<185 and x["weight_kg"]<80:
        return "Middle weight"
    else:
        return "Heavy weight"

bios["weight_category"]=bios.apply(categorize_atheletes, axis=1)
```
